**a. Ask the user to input two numbers, add them together and output (display) the result**
**1. Problem Definition**

INPUTS
first_num
second_num

PROCESS
total<-first_num + second_num

OUTPUT
total

**2. Answer**

START
    INPUT first_num
    INPUT second_num

    SET total<-first_num + second_num

    OUTPUT total
END
